## A. VISAYAS COMMUNICATION DATAFRAME

## C. CATEGORY-AVERAGE VISUALIZATION

### A.

```python
trackmean = board.groupby('Track')['Average'].mean().round(2)
gendermean = board.groupby('Gender')['Average'].mean().round(2)
hometownmean = board.groupby('Hometown')['Average'].mean().round(2)
```

### B.

```python
trackmean
```

```python
gendermean
```

```python
hometownmean
```

### C.

```python
plt.subplots(1,3, figsize=(16, 5))

plt.subplot(1,3,1)
plt.bar(trackmean.index, trackmean.values)
plt.ylim(0,100)
plt.title('Mean Average by Track')
plt.subplot(1,3,2)
plt.bar(gendermean.index, gendermean.values)
plt.ylim(0,100)
plt.title('Mean Average by Gender')
plt.subplot(1,3,3)
plt.bar(hometownmean.index, hometownmean.values)
plt.ylim(0,100)
plt.title('Mean Average by Hometown')
```

### D.

1. Track: Among the three tracks, Communication has the highest mean average track which is 67.97, then Microelectronics with 67.50 and Instrumentation having the lowest with 65.22.
2. Gender: Male has a higher mean average with 67.18 compared to the Female which is 66.62.
3. Hometown: Luzon has the highest mean average consisting of 68.08, followed by Mindanao which is 66.68 and with the least mean average is Visayas with 65.75
